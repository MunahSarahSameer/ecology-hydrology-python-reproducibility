```python

from pathlib import Path

ROOT = Path.cwd()

folders = [
    "repositories",
    "logs",
    "results",
    "scripts",
    "documentation",
    "figures",
    "manuscript"
]

for folder in folders:
    (ROOT / folder).mkdir(exist_ok=True)

print("Research workspace:", ROOT)
print("All research folders created!")
```
