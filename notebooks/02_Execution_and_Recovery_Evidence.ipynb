```python
from pathlib import Path
from datetime import datetime, timezone
import zipfile
import ast
import json

root = Path.cwd()
archive_path = root / "repositories" / "E01" / "code.zip"
results = root / "results"

print("=== STEP 187: E01 SAFE EXECUTION TARGET DISCOVERY ===")

targets = [
    "code/deepSDM/data_preparation/consolidate_observations.py",
    "code/deepSDM/data_preparation/create_splits.py",
    "code/deepSDM/util/helpers.py",
    "code/deepSDM/util/common.py",
    "code/deepSDM/util/schedulers.py"
]

report = {
    "candidate": "E01",
    "step": 187,
    "execution_attempted": False,
    "targets": [],
    "generated_at_utc": datetime.now(timezone.utc).isoformat()
}

with zipfile.ZipFile(archive_path) as archive:

    for filename in targets:

        print("\n" + "=" * 60)
        print("FILE:", filename)

        if filename not in archive.namelist():
            print("NOT FOUND")
            continue

        source = archive.read(filename).decode(
            "utf-8-sig",
            errors="replace"
        )

        tree = ast.parse(source, filename=filename)

        imports = []
        functions = []
        main_guard = False

        for node in tree.body:

            if isinstance(node, ast.Import):
                imports.extend(
                    alias.name for alias in node.names
                )

            elif isinstance(node, ast.ImportFrom):
                imports.append(
                    node.module or "(relative import)"
                )

            elif isinstance(
                node, (ast.FunctionDef, ast.AsyncFunctionDef)
            ):
                functions.append(node.name)

            elif isinstance(node, ast.If):
                if (
                    isinstance(node.test, ast.Compare)
                    and isinstance(node.test.left, ast.Name)
                    and node.test.left.id == "__name__"
                ):
                    main_guard = True

        print("Imports:", ", ".join(imports[:25]) or "None")
        print("Functions:", ", ".join(functions[:25]) or "None")
        print("Main guard detected:", main_guard)

        report["targets"].append({
            "file": filename,
            "imports": imports,
            "functions": functions,
            "main_guard_detected": main_guard
        })

output = results / "E01_step187_execution_target_discovery.json"

output.write_text(
    json.dumps(report, indent=2),
    encoding="utf-8"
)

print("\nEvidence saved:", output)
print("=== STEP 187 COMPLETE ===")
```

```python
from pathlib import Path
from datetime import datetime, timezone
import zipfile
import tempfile
import subprocess
import json
import sys

root = Path.cwd()
results = root / "results"

archive_path = (
    root / "repositories" / "H06" /
    "S4D_rainfall_runoff_simulations.zip"
)

target = (
    "S4D_rainfall_runoff_simulations-main/"
    "analysis/performance_functions.py"
)

print("=== STEP 195: H06 ORIGINAL METRIC EXECUTION ===")

with tempfile.TemporaryDirectory() as temp:
    temp = Path(temp)

    # Copy the exact original file from the archive.
    with zipfile.ZipFile(archive_path) as archive:
        original = archive.read(target)

    module_path = temp / "performance_functions.py"
    module_path.write_bytes(original)

    test_script = r'''
import importlib.util
import json
import numpy as np
import pandas as pd

spec = importlib.util.spec_from_file_location(
    "original_h06_metrics",
    "performance_functions.py"
)
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)

checks = {}

def record(name, actual, expected):
    checks[name] = {
        "actual": float(actual),
        "expected": float(expected),
        "pass": bool(np.isclose(
            actual, expected,
            rtol=1e-10,
            atol=1e-10
        ))
    }

# Test 1: Perfect simulated discharge.
perfect = pd.DataFrame({
    "qobs": [1.0, 2.0, 3.0, 4.0],
    "qsim": [1.0, 2.0, 3.0, 4.0]
})

record("perfect_nse", module.nse(perfect), 1.0)
record("perfect_kge", module.kge(perfect)[0], 1.0)
record("perfect_bias", module.bias(perfect), 0.0)

# Test 2: Imperfect predictions.
imperfect = pd.DataFrame({
    "qobs": [1.0, 2.0, 3.0, 4.0],
    "qsim": [2.0, 2.0, 2.0, 4.0]
})

obs = np.array([1., 2., 3., 4.])
sim = np.array([2., 2., 2., 4.])

expected_nse = 1 - (
    np.sum((sim - obs) ** 2)
    / np.sum((obs - obs.mean()) ** 2)
)

expected_bias = (
    (obs.mean() - sim.mean())
    / obs.mean() * 100
)

r = np.corrcoef(sim, obs)[0, 1]
alpha = np.std(sim) / np.std(obs)
beta = np.mean(sim) / np.mean(obs)

expected_kge = 1 - np.sqrt(
    (r - 1) ** 2
    + (alpha - 1) ** 2
    + (beta - 1) ** 2
)

record(
    "imperfect_nse",
    module.nse(imperfect),
    expected_nse
)
record(
    "imperfect_kge",
    module.kge(imperfect)[0],
    expected_kge
)
record(
    "imperfect_bias",
    module.bias(imperfect),
    expected_bias
)

# Test 3: Negative values must be excluded.
filtered = pd.DataFrame({
    "qobs": [1., 2., 3., 4., -1.],
    "qsim": [1., 2., 3., 4., 10.]
})

record(
    "negative_value_filtering",
    module.nse(filtered),
    1.0
)

# Test 4: Seed-specific simulation column.
seeded = pd.DataFrame({
    "qobs": [1., 2., 3., 4.],
    "qsim_7": [1., 2., 3., 4.]
})

record(
    "seed_specific_column",
    module.nse(seeded, seed=7),
    1.0
)

print(json.dumps({
    "checks": checks,
    "passed": sum(
        item["pass"] for item in checks.values()
    ),
    "total": len(checks)
}))
'''

    (temp / "test_runner.py").write_text(
        test_script, encoding="utf-8"
    )

    try:
        completed = subprocess.run(
            [sys.executable, "test_runner.py"],
            cwd=temp,
            capture_output=True,
            text=True,
            timeout=30
        )

        return_code = completed.returncode
        timed_out = False
        stdout = completed.stdout
        stderr = completed.stderr

    except subprocess.TimeoutExpired as error:
        return_code = None
        timed_out = True
        stdout = ""
        stderr = str(error)

    report = {
        "candidate": "H06",
        "step": 195,
        "source": target,
        "original_source_modified": False,
        "dependencies_installed": False,
        "return_code": return_code,
        "timed_out": timed_out,
        "stdout": stdout[-10000:],
        "stderr": stderr[-5000:],
        "generated_at_utc": datetime.now(
            timezone.utc
        ).isoformat()
    }

    if return_code == 0:
        try:
            parsed = json.loads(stdout)
            report.update(parsed)
            report["outcome"] = (
                "PASS"
                if parsed["passed"] == parsed["total"]
                else "FAIL"
            )
        except (ValueError, KeyError):
            report["outcome"] = "OUTPUT_PARSE_ERROR"
    else:
        report["outcome"] = (
            "TIMEOUT" if timed_out else "EXECUTION_FAILED"
        )

output = results / "H06_step195_metric_execution.json"

output.write_text(
    json.dumps(report, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

print("\nRETURN CODE:", report["return_code"])
print("TIMED OUT:", report["timed_out"])

if "checks" in report:
    print("\n=== VALIDATION CHECKS ===")
    for name, item in report["checks"].items():
        print(
            "PASS" if item["pass"] else "FAIL",
            name,
            "| actual:", round(item["actual"], 8),
            "| expected:", round(item["expected"], 8)
        )

print("\nSTDERR:")
print(report["stderr"][:2500] or "(empty)")

print("\nOUTCOME:", report["outcome"])

if "passed" in report:
    print(
        "Checks:",
        report["passed"],
        "/",
        report["total"]
    )

print("Evidence saved:", output)
print("=== STEP 195 COMPLETE ===")
```

```python
from pathlib import Path
from datetime import datetime, timezone
import zipfile
import io
import pandas as pd
import numpy as np
import json
import re

root = Path.cwd()
results = root / "results"

archive_path = (
    root / "repositories" / "H01" / "Data and codes.zip"
)

previous_path = results / "H01_step201_pandas_recovery.json"

print("=== STEP 202: H01 INDEPENDENT NSE VALIDATION ===")

previous = json.loads(
    previous_path.read_text(encoding="utf-8")
)

assert previous["return_code"] == 0

with zipfile.ZipFile(archive_path) as archive:

    def load_workbook(filename):
        member = "Data and codes/data/" + filename

        df = pd.read_excel(
            io.BytesIO(archive.read(member)),
            index_col=0
        )

        df.index = pd.to_datetime(df.index)

        return df

    natural = load_workbook("monthly_streamflow_NAT.xlsx")
    human = load_workbook("monthly_streamflow_HUM.xlsx")
    observed = load_workbook("monthly_streamflow_OBS.xlsx")

def independent_nse(obs, sim):

    valid = (
        np.isfinite(obs)
        & np.isfinite(sim)
    )

    obs = np.asarray(obs)[valid]
    sim = np.asarray(sim)[valid]

    if len(obs) < 2:
        return float("nan")

    denominator = np.sum(
        (obs - np.mean(obs)) ** 2
    )

    if denominator == 0:
        return float("nan")

    return float(
        1 - np.sum((sim - obs) ** 2) / denominator
    )

periods = [
    ("1991-2000", "1991", "2000"),
    ("2001-2012", "2001", "2012"),
    ("2013-2022", "2013", "2022")
]

stations = ["xiantao", "yichang", "datong"]

pattern = re.compile(
    r"(\w+) \((\d{4}-\d{4})\): "
    r"NAT=([-+0-9.eE]+), HUM=([-+0-9.eE]+)"
)

printed = {}

for match in pattern.finditer(previous["stdout"]):

    station, period, nat, hum = match.groups()

    printed[(station, period)] = {
        "NAT": float(nat),
        "HUM": float(hum)
    }

checks = []

for station in stations:

    for period, start, end in periods:

        obs = observed.loc[start:end, station]
        nat = natural.loc[start:end, station]
        hum = human.loc[start:end, station]

        expected_nat = independent_nse(
            obs.to_numpy(),
            nat.to_numpy()
        )

        expected_hum = independent_nse(
            obs.to_numpy(),
            hum.to_numpy()
        )

        original = printed.get((station, period))

        for label, expected in [
            ("NAT", expected_nat),
            ("HUM", expected_hum)
        ]:

            actual = (
                original[label]
                if original is not None
                else None
            )

            passed = (
                actual is not None
                and bool(np.isclose(
                    actual,
                    expected,
                    rtol=1e-8,
                    atol=1e-8
                ))
            )

            checks.append({
                "station": station,
                "period": period,
                "scenario": label,
                "original_output": actual,
                "independent_calculation": expected,
                "pass": passed
            })

report = {
    "candidate": "H01",
    "step": 202,
    "test_type": "independent_nse_validation",
    "checks": checks,
    "passed": sum(item["pass"] for item in checks),
    "total": len(checks),
    "generated_at_utc": datetime.now(
        timezone.utc
    ).isoformat()
}

report["outcome"] = (
    "PASS"
    if report["passed"] == report["total"]
    else "FAIL"
)

output = results / "H01_step202_nse_validation.json"

output.write_text(
    json.dumps(report, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

for item in checks:

    print(
        "PASS" if item["pass"] else "FAIL",
        item["station"],
        item["period"],
        item["scenario"],
        "| original:", item["original_output"],
        "| independent:", round(
            item["independent_calculation"], 8
        )
    )

print("\nRESULT:", report["outcome"])
print("Checks:", report["passed"], "/", report["total"])
print("Evidence saved:", output)
print("=== STEP 202 COMPLETE ===")
```

```python
from pathlib import Path
from datetime import datetime, timezone
from collections import Counter
import json
import csv

root = Path.cwd()
results = root / "results"

print("=== STEP 213: VERIFIED EVIDENCE MATRIX ===")

checklist = json.loads(
    (results / "testing_checklist_15_candidates.json")
    .read_text(encoding="utf-8")
)

records = checklist["records"]

assert len(records) == 15
assert len({r["candidate_id"] for r in records}) == 15

matrix = []

for record in records:

    candidate_id = record["candidate_id"]

    domain = (
        "Ecology" if candidate_id.startswith("E")
        else "Hydrology" if candidate_id.startswith("H")
        else "Unknown"
    )

    evidence = [
        item.strip()
        for item in record.get(
            "evidence_files_found", ""
        ).split(";")
        if item.strip()
    ]

    missing = []

    for item in evidence:
        path = Path(item.replace("\\", "/"))

        if not path.is_absolute():
            path = root / path

        if not path.is_file():
            missing.append(item)

    row = {
        "candidate_id": candidate_id,
        "domain": domain,
        "baseline_status": record.get(
            "baseline_status", "UNKNOWN"
        ),
        "later_test_status": record.get(
            "recovered_status", "UNKNOWN"
        ),
        "testing_review_status": record.get(
            "testing_review_status", ""
        ),
        "full_scientific_workflow_verified": record.get(
            "full_scientific_workflow_verified",
            "NOT_VERIFIED"
        ),
        "published_results_reproduced": record.get(
            "published_results_reproduced",
            "NOT_VERIFIED"
        ),
        "evidence_reference_count": len(evidence),
        "missing_evidence_count": len(missing),
        "evidence_files": "; ".join(evidence),
        "baseline_interpretation": record.get(
            "baseline_interpretation", ""
        ),
        "component_execution_evidence": record.get(
            "component_execution_evidence", ""
        ),
        "previous_finding": record.get(
            "previous_finding", ""
        ),
        "next_action": record.get(
            "next_action", ""
        )
    }

    matrix.append(row)

baseline_counts = Counter(
    row["baseline_status"] for row in matrix
)

later_counts = Counter(
    row["later_test_status"] for row in matrix
)

missing_total = sum(
    row["missing_evidence_count"] for row in matrix
)

print("\nBASELINE COUNTS")
for status, count in sorted(baseline_counts.items()):
    print(f"{status}: {count}")

print("\nLATER TEST COUNTS")
for status, count in sorted(later_counts.items()):
    print(f"{status}: {count}")

print("\nCANDIDATE MATRIX")
print("-" * 85)

for row in matrix:
    print(
        f"{row['candidate_id']:4} | "
        f"{row['domain']:10} | "
        f"Baseline: {row['baseline_status']:14} | "
        f"Later: {row['later_test_status']:14} | "
        f"Evidence: {row['evidence_reference_count']}"
    )

report = {
    "step": 213,
    "generated_at_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "candidate_count": len(matrix),
    "baseline_counts": dict(baseline_counts),
    "later_test_counts": dict(later_counts),
    "missing_evidence_references": missing_total,
    "interpretation_limits": [
        "PASS and PASS_LIMITED are not equivalent "
        "to full scientific workflow reproduction.",
        "Later successful component tests do not "
        "necessarily represent baseline success.",
        "A timeout alone does not establish code decay.",
        "Recovery effort cannot be quantified in minutes "
        "without recorded timing evidence.",
        "The 15 candidates are not a representative "
        "sample of all ecology and hydrology software."
    ],
    "matrix": matrix
}

json_path = results / "step213_verified_evidence_matrix.json"
csv_path = results / "step213_verified_evidence_matrix.csv"

json_path.write_text(
    json.dumps(
        report,
        indent=2,
        ensure_ascii=False
    ),
    encoding="utf-8"
)

with csv_path.open(
    "w",
    newline="",
    encoding="utf-8-sig"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=list(matrix[0].keys())
    )

    writer.writeheader()
    writer.writerows(matrix)

print("\nMissing evidence references:", missing_total)

print("\nFiles saved:")
print(json_path)
print(csv_path)

print("\n=== STEP 213 COMPLETE ===")
```

```python
from pathlib import Path
from datetime import date, datetime, timezone
import json
import csv

root = Path.cwd()
results = root / "results"

print("=== STEP 215: VERIFIED ARCHIVE AGE CALCULATION ===")

reference_date = date(2026, 11, 12)

matrix = json.loads(
    (results / "step213_verified_evidence_matrix.json")
    .read_text(encoding="utf-8")
)["matrix"]

with (
    results / "consolidated_verified_provenance.csv"
).open(encoding="utf-8-sig", newline="") as file:
    provenance = list(csv.DictReader(file))

by_id = {
    row["package_id"]: row
    for row in provenance
}

report = {
    "step": 215,
    "reference_date": reference_date.isoformat(),
    "generated_at_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "statistical_test_performed": False,
    "candidates": []
}

print("\nVERIFIED ARCHIVE AGE TABLE")
print("-" * 85)

for record in matrix:

    cid = record["candidate_id"]
    baseline = record["baseline_status"]

    source = by_id.get(cid, {})

    verified = (
        source.get("provenance_verification_status")
        == "ZENODO_CHECKSUM_MATCH"
    )

    date_string = (
        source.get("verified_archive_date", "")
        if verified else ""
    )

    archive_age_days = None
    archive_age_years = None

    if date_string:
        archive_date = date.fromisoformat(date_string)

        archive_age_days = (
            reference_date - archive_date
        ).days

        archive_age_years = round(
            archive_age_days / 365.2425, 3
        )

    eligible = (
        verified
        and archive_age_days is not None
        and archive_age_days >= 0
        and baseline in {"PASS", "FAIL"}
    )

    item = {
        "candidate_id": cid,
        "domain": record["domain"],
        "baseline_status": baseline,
        "verified_archive_date": date_string or None,
        "archive_age_days": archive_age_days,
        "archive_age_years": archive_age_years,
        "verified_provenance": verified,
        "potentially_eligible": eligible,
        "scope_comparability_verified": False
    }

    report["candidates"].append(item)

    age_display = (
        f"{archive_age_years:.3f}"
        if archive_age_years is not None
        else "UNKNOWN"
    )

    print(
        f"{cid:4} | "
        f"Age: {age_display:8} years | "
        f"Baseline: {baseline:14} | "
        f"Eligible: {eligible}"
    )

eligible_records = [
    row for row in report["candidates"]
    if row["potentially_eligible"]
]

report["potentially_eligible_count"] = len(
    eligible_records
)

report["limitations"] = [
    "Archive age is measured from verified Zenodo archive dates.",
    "Only PASS/FAIL baselines are potentially eligible.",
    "Baseline testing scopes have not yet been established as comparable.",
    "The sample is small and purposively selected.",
    "No statistical association or causal relationship is claimed."
]

json_path = results / "step215_verified_archive_ages.json"
csv_path = results / "step215_verified_archive_ages.csv"

json_path.write_text(
    json.dumps(
        report,
        indent=2,
        ensure_ascii=False
    ),
    encoding="utf-8"
)

with csv_path.open(
    "w",
    newline="",
    encoding="utf-8-sig"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=list(report["candidates"][0].keys())
    )

    writer.writeheader()
    writer.writerows(report["candidates"])

print("\nPotentially eligible candidates:", len(
    eligible_records
))

print("\nEvidence saved:")
print(json_path)
print(csv_path)

print("\n=== STEP 215 COMPLETE ===")
```

```python
from pathlib import Path
from datetime import datetime, timezone
import json

root = Path.cwd()
results = root / "results"

print("=== STEP 218: BARRIER EVIDENCE VERIFICATION ===")

classification_path = (
    results / "step217_execution_barrier_classification.json"
)

classification = json.loads(
    classification_path.read_text(encoding="utf-8")
)

# Existing evidence selected for verification.
# We will inspect actual JSON structures before making
# any further classification changes.

evidence_targets = {
    "E07": [
        "repositories/results/E07_consolidated_evidence.json"
    ],
    "H01": [
        "results/H01_step200_path_recovery.json",
        "results/H01_step201_pandas_recovery.json",
        "results/H01_step202_nse_validation.json"
    ],
    "H02": [
        "repositories/results/H02_test1_validation.json"
    ],
    "H03": [
        "results/H03_step205_workload_estimate.json",
        "results/H03_step207_bounded_execution.json"
    ],
    "H05": [
        "repositories/results/H05_pandas_control_summary.json"
    ]
}

report = {
    "step": 218,
    "generated_at_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "classification_modified": False,
    "original_archives_modified": False,
    "candidates": {}
}

for record in classification["records"]:

    cid = record["candidate_id"]

    print("\n" + "=" * 65)
    print("CANDIDATE:", cid)
    print("CLASSIFIED BARRIERS:", record["observed_barriers"])
    print("CLASSIFIED INTERVENTIONS:", record["interventions"])

    candidate_evidence = []

    for relative_path in evidence_targets.get(cid, []):

        path = root / relative_path

        item = {
            "path": relative_path,
            "found": path.is_file()
        }

        print("\nEVIDENCE:", relative_path)
        print("Found:", item["found"])

        if path.is_file():

            try:
                data = json.loads(
                    path.read_text(encoding="utf-8-sig")
                )

                item["json_valid"] = True
                item["top_level_keys"] = (
                    list(data.keys())
                    if isinstance(data, dict)
                    else []
                )

                print("JSON valid: YES")
                print("Top-level keys:", item["top_level_keys"])

                # Show compact evidence without assuming
                # that all files use the same JSON schema.
                if isinstance(data, dict):

                    summary = {}

                    for key, value in data.items():

                        if (
                            isinstance(value, (str, int, float, bool))
                            or value is None
                        ):
                            summary[key] = value

                        elif isinstance(value, dict):

                            summary[key] = {
                                k: v
                                for k, v in value.items()
                                if (
                                    isinstance(
                                        v, (str, int, float, bool)
                                    )
                                    or v is None
                                )
                            }

                    item["summary"] = summary

                    print(
                        json.dumps(
                            summary,
                            indent=2,
                            ensure_ascii=False
                        )[:3500]
                    )

            except Exception as error:

                item["json_valid"] = False
                item["error"] = str(error)

                print("JSON reading error:", error)

        candidate_evidence.append(item)

    report["candidates"][cid] = {
        "classified_barriers": record["observed_barriers"],
        "classified_interventions": record["interventions"],
        "evidence": candidate_evidence
    }

missing = [
    item["path"]
    for candidate in report["candidates"].values()
    for item in candidate["evidence"]
    if not item["found"]
]

report["missing_evidence_files"] = missing
report["verification_status"] = "AWAITING_MANUAL_REVIEW"

output = results / "step218_barrier_evidence_verification.json"

output.write_text(
    json.dumps(
        report,
        indent=2,
        ensure_ascii=False
    ),
    encoding="utf-8"
)

print("\n" + "=" * 65)
print("Candidates reviewed:", len(report["candidates"]))
print("Missing evidence files:", len(missing))

for filename in missing:
    print("MISSING:", filename)

print("Classifications modified: NO")
print("Original archives modified: NO")
print("Evidence saved:", output)
print("=== STEP 218 COMPLETE ===")
```

```python
from pathlib import Path
from datetime import datetime, timezone
import json
import csv

root = Path.cwd()
results = root / "results"

print("=== STEP 224: UNRESOLVED EVIDENCE AUDIT ===")

matrix = json.loads(
    (results / "step223_complete_classification_matrix.json")
    .read_text(encoding="utf-8")
)

targets = ["E06", "E08"]

keywords = (
    "E06", "E08", "recovery", "reproduction",
    "dependency", "baseline", "comparison",
    "notebook", "provenance"
)

def inspect_json(path):
    try:
        data = json.loads(
            path.read_text(encoding="utf-8-sig")
        )

        if isinstance(data, dict):
            return {
                "type": "dict",
                "keys": list(data.keys())[:35],
                "relevant_values": {
                    k: str(v)[:500]
                    for k, v in data.items()
                    if any(
                        word in k.lower()
                        for word in (
                            "outcome", "status", "test",
                            "intervention", "scope",
                            "limitation", "reproduc",
                            "comparison", "modified"
                        )
                    )
                }
            }

        return {
            "type": type(data).__name__,
            "item_count": (
                len(data)
                if hasattr(data, "__len__")
                else None
            ),
            "excerpt": str(data)[:600]
        }

    except Exception as exc:
        return {"read_error": str(exc)}

report = {
    "step": 224,
    "generated_at_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "original_archives_modified": False,
    "classification_matrix_modified": False,
    "targets": []
}

for cid in targets:

    record = next(
        r for r in matrix["records"]
        if r["candidate_id"] == cid
    )

    print("\n" + "=" * 70)
    print("CANDIDATE:", cid)
    print("Baseline:", record["baseline_status"])
    print("Later:", record["later_test_status"])
    print("Current scope:", record["tested_scope"])
    print("Current barriers:", record["observed_barriers"])

    matching_files = sorted({
        p for directory in (
            results,
            root / "repositories" / "results"
        )
        if directory.exists()
        for p in directory.rglob("*")
        if p.is_file()
        and cid.lower() in p.name.lower()
        and p.suffix.lower() in (".json", ".txt", ".log")
    })

    entry = {
        "candidate_id": cid,
        "files": []
    }

    for path in matching_files:

        relative = str(path.relative_to(root))

        print("\nEvidence:", relative)

        file_record = {
            "path": relative,
            "size_bytes": path.stat().st_size
        }

        if path.suffix.lower() == ".json":

            detail = inspect_json(path)
            file_record["inspection"] = detail

            print(
                json.dumps(
                    detail,
                    indent=2,
                    ensure_ascii=False
                )[:1800]
            )

        else:

            content = path.read_text(
                encoding="utf-8-sig",
                errors="replace"
            )

            file_record["excerpt"] = content[:1200]

            print(content[:1200])

        entry["files"].append(file_record)

    report["targets"].append(entry)

output = results / "step224_unresolved_evidence_audit.json"

output.write_text(
    json.dumps(
        report,
        indent=2,
        ensure_ascii=False
    ),
    encoding="utf-8"
)

print("\n" + "=" * 70)
print("Candidates audited:", len(report["targets"]))
print(
    "Evidence files inspected:",
    sum(len(x["files"]) for x in report["targets"])
)
print("Original archives modified: NO")
print("Classification matrix modified: NO")
print("Saved:", output)
print("=== STEP 224 COMPLETE ===")
```

```python
from pathlib import Path
from collections import Counter
from datetime import datetime, timezone
import json

root = Path.cwd()
results = root / "results"

print("=== STEP 225: METHODOLOGY AND EVIDENCE-INTEGRITY AUDIT ===")

matrix = json.loads(
    (results / "step223_complete_classification_matrix.json")
    .read_text(encoding="utf-8")
)

provenance_path = results / "consolidated_verified_provenance.csv"

import csv

with provenance_path.open(
    encoding="utf-8-sig",
    newline=""
) as file:
    provenance = list(csv.DictReader(file))

provenance_lookup = {
    r["package_id"]: r
    for r in provenance
}

records = matrix["records"]

assert len(records) == 15
assert len({r["candidate_id"] for r in records}) == 15

issues = []
candidate_audit = []

for row in records:

    cid = row["candidate_id"]
    prov = provenance_lookup.get(cid)

    if prov is None:
        issues.append(f"{cid}: Missing provenance row")
        continue

    age = row["archive_age_years"]

    verified_age = (
        age not in ("", None, "UNKNOWN", "NOT_VERIFIED")
        and str(age).strip() != ""
    )

    baseline = row["baseline_status"]

    comparable_binary_status = baseline in ("PASS", "FAIL")

    eligible_for_age_description = verified_age

    eligible_for_exploratory_age_outcome = (
        verified_age and comparable_binary_status
    )

    audit = {
        "candidate_id": cid,
        "domain": row["domain"],
        "baseline_status": baseline,
        "later_test_status": row["later_test_status"],
        "review_level": row["review_level"],
        "verified_archive_age_available": verified_age,
        "binary_baseline_available": comparable_binary_status,
        "potential_age_outcome_candidate":
            eligible_for_exploratory_age_outcome,
        "full_workflow_verified": row["full_workflow_verified"],
        "published_results_reproduced":
            row["published_results_reproduced"],
        "missing_evidence_count":
            int(row["missing_evidence_count"])
    }

    candidate_audit.append(audit)

    if audit["missing_evidence_count"] != 0:
        issues.append(f"{cid}: Missing evidence references")

baseline_counts = Counter(
    r["baseline_status"] for r in records
)

domain_counts = Counter(
    r["domain"] for r in records
)

review_counts = Counter(
    r["review_level"] for r in records
)

verified_ages = sum(
    r["verified_archive_age_available"]
    for r in candidate_audit
)

potential_age_candidates = [
    r["candidate_id"]
    for r in candidate_audit
    if r["potential_age_outcome_candidate"]
]

report = {
    "step": 225,
    "generated_at_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "study_design": "PURPOSIVE_EXPLORATORY_CASE_STUDY",
    "candidate_count": len(records),
    "domain_counts": dict(domain_counts),
    "baseline_counts": dict(baseline_counts),
    "review_counts": dict(review_counts),
    "verified_archive_ages": verified_ages,
    "potential_age_outcome_candidates":
        potential_age_candidates,
    "potential_age_outcome_count":
        len(potential_age_candidates),
    "age_analysis_warning": (
        "Binary baseline labels do not establish comparable "
        "execution tests. No correlation or causal "
        "relationship should be inferred."
    ),
    "human_recovery_minutes_measured": False,
    "full_workflow_reproduction_verified": False,
    "published_paper_reproduction_verified": False,
    "original_archives_modified": False,
    "methodological_issues": issues,
    "candidate_audit": candidate_audit
}

output = results / "step225_methodology_integrity_audit.json"

output.write_text(
    json.dumps(report, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

print("\nStudy design:", report["study_design"])
print("Candidates:", report["candidate_count"])
print("Domains:", report["domain_counts"])
print("Baseline outcomes:", report["baseline_counts"])
print("Review levels:", report["review_counts"])
print("Verified archive ages:", verified_ages)
print(
    "Potential age-outcome candidates:",
    potential_age_candidates
)
print("Methodological issues:", issues)
print("Human recovery minutes measured: NO")
print("Full scientific reproduction verified: NO")
print("Saved:", output)

print("\n=== STEP 225 COMPLETE ===")
```

```python
from pathlib import Path
from collections import Counter
from datetime import datetime, timezone
import csv
import json

root = Path.cwd()
results = root / "results"
results.mkdir(exist_ok=True)

print("=== STEP 226: PUBLICATION RESULTS TABLES ===")

matrix = json.loads(
    (results / "step223_complete_classification_matrix.json")
    .read_text(encoding="utf-8")
)

records = matrix["records"]

assert len(records) == 15
assert len({r["candidate_id"] for r in records}) == 15

def write_csv(filename, rows):
    path = results / filename
    with path.open(
        "w", encoding="utf-8-sig", newline=""
    ) as f:
        writer = csv.DictWriter(
            f, fieldnames=list(rows[0].keys())
        )
        writer.writeheader()
        writer.writerows(rows)
    return path

# TABLE 1: Sample and historical baseline outcomes

table1 = []
for domain in ("Ecology", "Hydrology"):
    subset = [
        r for r in records if r["domain"] == domain
    ]
    counts = Counter(
        r["baseline_status"] for r in subset
    )
    table1.append({
        "domain": domain,
        "repositories": len(subset),
        "baseline_pass": counts["PASS"],
        "baseline_fail": counts["FAIL"],
        "baseline_unknown": counts["UNKNOWN"],
        "baseline_not_attempted": counts["NOT_ATTEMPTED"]
    })

# TABLE 2: Individual execution and recovery findings

table2 = []
for r in records:
    table2.append({
        "candidate_id": r["candidate_id"],
        "domain": r["domain"],
        "baseline_status": r["baseline_status"],
        "later_test_status": r["later_test_status"],
        "tested_scope": r["tested_scope"],
        "observed_barriers": r["observed_barriers"],
        "documented_interventions": r["interventions"],
        "review_level": r["review_level"],
        "qualification": r["qualification"]
    })

# TABLE 3: Archive ages and baseline eligibility

table3 = []
for r in records:
    age = r["archive_age_years"]
    has_age = age not in (None, "", "UNKNOWN", "NOT_VERIFIED")

    table3.append({
        "candidate_id": r["candidate_id"],
        "domain": r["domain"],
        "archive_age_years": age if has_age else "",
        "archive_age_available": has_age,
        "baseline_status": r["baseline_status"],
        "binary_baseline_available":
            r["baseline_status"] in ("PASS", "FAIL"),
        "comparable_test_scope_established": False
    })

# TABLE 4: Evidence and limitations

table4 = []
for r in records:
    table4.append({
        "candidate_id": r["candidate_id"],
        "review_level": r["review_level"],
        "evidence_reference_count":
            r["evidence_reference_count"],
        "missing_evidence_count":
            r["missing_evidence_count"],
        "full_workflow_verified":
            r["full_workflow_verified"],
        "published_results_reproduced":
            r["published_results_reproduced"],
        "qualification": r["qualification"]
    })

paths = [
    write_csv("step226_table1_sample_baseline.csv", table1),
    write_csv("step226_table2_execution_recovery.csv", table2),
    write_csv("step226_table3_archive_age.csv", table3),
    write_csv("step226_table4_evidence_limitations.csv", table4)
]

baseline = Counter(
    r["baseline_status"] for r in records
)
later = Counter(
    r["later_test_status"] for r in records
)
review = Counter(
    r["review_level"] for r in records
)

summary = {
    "step": 226,
    "generated_at_utc":
        datetime.now(timezone.utc).isoformat(),
    "sample_size": len(records),
    "domain_counts": dict(Counter(
        r["domain"] for r in records
    )),
    "baseline_counts": dict(baseline),
    "later_status_counts": dict(later),
    "review_counts": dict(review),
    "verified_archive_age_count": sum(
        x["archive_age_available"] for x in table3
    ),
    "binary_baseline_and_age_count": sum(
        x["archive_age_available"]
        and x["binary_baseline_available"]
        for x in table3
    ),
    "full_scientific_reproductions_verified": 0,
    "human_recovery_time_measured": False,
    "age_failure_correlation_established": False,
    "tables": [str(p.relative_to(root)) for p in paths],
    "limitations": [
        "Purposive sample; no population generalisation.",
        "Baseline test scopes are heterogeneous.",
        "Seven baselines are unknown or not attempted.",
        "Nine barrier classifications remain provisional.",
        "Dependency absence does not prove temporal decay.",
        "No full published-paper reproduction verified.",
        "Human recovery time not reliably measured."
    ]
}

summary_path = results / "step226_results_summary.json"
summary_path.write_text(
    json.dumps(summary, indent=2, ensure_ascii=False),
    encoding="utf-8"
)

print("\nSAMPLE")
for row in table1:
    print(row)

print("\nBASELINE:", dict(baseline))
print("LATER OUTCOMES:", dict(later))
print("REVIEW LEVELS:", dict(review))
print("ARCHIVE AGES AVAILABLE:",
      summary["verified_archive_age_count"])
print("AGE + BINARY BASELINE:",
      summary["binary_baseline_and_age_count"])

print("\nOUTPUT FILES")
for p in paths + [summary_path]:
    print(p)

print("\nOriginal archives modified: NO")
print("Existing evidence modified: NO")
print("=== STEP 226 COMPLETE ===")
```

```python
from pathlib import Path
from datetime import datetime, timezone
from hashlib import sha256
from zipfile import ZipFile, ZIP_DEFLATED
from collections import Counter
import json

root = Path.cwd()
results = root / "results"
backups = root / "backups"

backups.mkdir(exist_ok=True)

print("=== STEP 228: FINAL EVIDENCE FREEZE AND BACKUP ===")

required = [
    "step223_complete_classification_matrix.json",
    "step224_unresolved_evidence_audit.json",
    "step225_methodology_integrity_audit.json",
    "step226_results_summary.json",
    "step227_research_framework.json"
]

for name in required:
    path = results / name
    assert path.is_file(), f"Required evidence missing: {name}"

# Include research evidence and documentation, not original repositories.
folders = [
    results,
    root / "repositories" / "results"
]

files = sorted({
    p for folder in folders
    if folder.exists()
    for p in folder.rglob("*")
    if p.is_file()
    and p.suffix.lower() in (
        ".json", ".csv", ".txt", ".log", ".md"
    )
})

assert files, "No evidence files found"

def file_hash(path):
    digest = sha256()
    with path.open("rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()

manifest_files = []

for path in files:
    manifest_files.append({
        "relative_path": path.relative_to(root).as_posix(),
        "size_bytes": path.stat().st_size,
        "sha256": file_hash(path)
    })

timestamp = datetime.now(timezone.utc).strftime(
    "%Y%m%d_%H%M%S"
)

manifest = {
    "step": 228,
    "generated_at_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "project_title": (
        "Evaluating Component-Level Executability "
        "and Recovery Barriers in Archived Python "
        "Research Code from Ecology and Hydrology"
    ),
    "research_evidence_frozen": True,
    "original_repositories_included": False,
    "original_repositories_modified": False,
    "archive_scope": (
        "JSON, CSV, TXT, LOG and Markdown evidence "
        "under results/ and repositories/results/"
    ),
    "file_count": len(manifest_files),
    "files": manifest_files,
    "limitations": [
        "Backup excludes original repository archives.",
        "Backup does not independently verify scientific results.",
        "Some classifications remain provisional.",
        "Full scientific reproduction is unverified.",
        "Reference and novelty verification remain pending."
    ]
}

manifest_path = backups / (
    f"research_evidence_manifest_{timestamp}.json"
)

manifest_path.write_text(
    json.dumps(
        manifest, indent=2, ensure_ascii=False
    ),
    encoding="utf-8"
)

zip_path = backups / (
    f"research_evidence_frozen_{timestamp}.zip"
)

with ZipFile(
    zip_path, "w", compression=ZIP_DEFLATED
) as archive:

    for path in files:
        archive.write(
            path,
            arcname=path.relative_to(root).as_posix()
        )

    archive.write(
        manifest_path,
        arcname="research_evidence_manifest.json"
    )

# Verify every archived evidence file against its SHA-256.
with ZipFile(zip_path, "r") as archive:

    assert archive.testzip() is None

    for record in manifest_files:
        archived_bytes = archive.read(
            record["relative_path"]
        )
        actual_hash = sha256(archived_bytes).hexdigest()

        assert actual_hash == record["sha256"], (
            f"Checksum mismatch: {record['relative_path']}"
        )

print("\nEvidence files:", len(files))
print(
    "Total evidence size (MB):",
    round(sum(p.stat().st_size for p in files) / 1048576, 2)
)
print(
    "Backup size (MB):",
    round(zip_path.stat().st_size / 1048576, 2)
)
print("Checksum verification: ALL PASSED")
print("Original repository archives modified: NO")

print("\nMANIFEST:")
print(manifest_path)

print("\nVERIFIED ZIP BACKUP:")
print(zip_path)

print("\n=== STEP 228 COMPLETE ===")
```
